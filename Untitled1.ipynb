```python
!pip install reportlab -q

from reportlab.lib.pagesizes import A4
from reportlab.pdfgen import canvas
from google.colab import files

# PDF file name
file_name = "class_activities.pdf"

# A4 page
width, height = A4

# Create PDF
c = canvas.Canvas(file_name, pagesize=A4)

# -----------------------------
# DATE
# -----------------------------
c.setFont("Helvetica", 16)
c.drawString(365, 720, "Date : 07.10.2026")

# -----------------------------
# NAME
# -----------------------------
c.setFont("Helvetica", 14)

c.drawString(55, 665, "Name :")
c.line(95, 662, 245, 662)

c.drawString(250, 665, "Aaron Sebastian")
c.line(315, 662, 465, 662)

# -----------------------------
# SUBJECT
# -----------------------------
c.drawString(55, 625, "Subject :")
c.line(115, 622, 400, 622)

# -----------------------------
# CLASS ACTIVITIES
# -----------------------------
c.drawString(55, 575, "Class activities:")
c.line(55, 555, 540, 555)

# -----------------------------
# TASK 1
# -----------------------------
c.drawString(55, 525, "Task 1) List :")

x = '{} Nordic University'
y = x.format('Riga')

c.drawString(75, 495, y)

# -----------------------------
# SAVE PDF
# -----------------------------
c.save()

print("PDF created successfully!")

# -----------------------------
# DOWNLOAD PDF
# -----------------------------
files.download(file_name)
```
